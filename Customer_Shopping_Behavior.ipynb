```python
labels = ['Young', 'Adult', 'Middle-aged', 'Senior']
df['age_groups'] = pd.qcut(df['age'], q = 4, labels = labels)
```

```python
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 months' : 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
